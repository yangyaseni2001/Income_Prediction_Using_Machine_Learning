```python
import matplotlib.pyplot as plt

# Select numerical columns
numerical_cols_data = ['age', 'hours-per-week']

# Boxplot before outlier removal
plt.figure()
plt.boxplot([adult_data[col] for col in numerical_cols_data], labels=numerical_cols_data)
plt.title("Boxplot Before Outlier Removal")
plt.ylabel("Values")
plt.show()
```

```python
def remove_outliers(df,col):
   Q1= df[col].quantile(0.25)
   Q3= df[col].quantile(0.75)
   IQR= Q3-Q1
   lower = Q1- 1.5*IQR
   upper= Q3+ 1.5*IQR
   return df[(df[col]>=lower) & (df[col]<=upper)]
```

```python
for col in numerical_col_data:
   adult_data= remove_outliers(adult_data,col)
adult_data.head()
```

```python
plt.figure(figsize=(5, 5))
plt.boxplot([adult_data[col] for col in numerical_cols_data],
            labels=numerical_cols_data)
plt.title("Boxplot After Outlier Removal")
plt.ylabel("Values")
plt.show()
```

```python
adult_data['income']=adult_data['income'].str.strip().str.lower()
adult_data['income_binary']=adult_data['income'].map({'<=50k':0,'>50k':1})
adult_data.head(10)
```

```python
from scipy.stats import pointbiserialr

r,p= pointbiserialr(adult_data['age'],adult_data['income_binary'])
print('Correlation: ',r)
print('P-value: ', p)
```

```python
# select numerical columns exactly from adult_data
numerical_cols = adult_data.select_dtypes(include=['int64', 'float64'])

# calculate correlation
corr_matrix = numerical_cols.corr()

# print correlation values
print("Correlation Values:")
print(corr_matrix)
```

```python
from scipy.stats import pointbiserialr

features = ['age', 'education-num', 'hours-per-week']
# include fnlwgt or fnlweight depending on your column name
features.append('fnlwgt' if 'fnlwgt' in adult_data.columns else 'fnlweight')

for col in features:
   valid = adult_data[[col, 'income_binary']].dropna()
   r, p = pointbiserialr(valid[col], valid['income_binary'])
   print(f'{col}: Correlation = {r:.4f}, P-value = {p:.4e}')

 
```

```python
plt.figure(figsize=(5,5))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```
